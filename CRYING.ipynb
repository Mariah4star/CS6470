<h3 style="color:#275d38; margin:20px 0 6px 0;">Q4</h3>

Replace the loop above with a single expression giving the same number.

```python
v = np.array([1., 2., 3., 4.])
w = np.array([10., 20., 30., 40.])

total = 0.0
for i in range(len(v)):
    total += v[i] * w[i]

# Q4. The same number as the loop above, in one expression.
q4 = np.dot(v, w) # or q4 = (v * w).sum()

print(total, q4)
```
