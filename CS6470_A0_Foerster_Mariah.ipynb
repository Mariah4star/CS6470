## Task 2 of 6 — Load a small CSV

Default: public seaborn `tips` table (restaurant bills). You may paste your own public CSV URL or a tiny table you created. Workplace data is **not** required.


```python
# Default public CSV — this cell runs as-is.
# (optional): replace csv_url with another public CSV.
csv_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(csv_url)
print("Loaded rows:", len(df), "from", csv_url)

# (optional alternative): comment out the two lines above and load a pasted table:
#from io import StringIO
#df = pd.read_csv(StringIO("""total_bill,tip,size
#16.99,1.01,2
#10.34,1.66,3
#21.01,3.50,3
#"""))

#print(df)
```

## Task 4 of 6 — Classification target and train/test split

Build a simple yes/no target with a **median split** on a numeric column, then call `train_test_split` with a fixed `random_state` so the split is reproducible.


```python
from sklearn.model_selection import train_test_split

# TODO (optional): if you used your own CSV, pick a numeric column for the
# median split and one or two numeric feature columns.
target_col = "tip"
feature_cols = ["total_bill", "size"]

y = (df[target_col] >= df[target_col].median()).astype(int)
X = df[feature_cols]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print("X_train", X_train.shape, "X_test", X_test.shape)
print("class rate in y (share of 1s):", float(y.mean()))
```

## Task 5 of 6 — Fit DummyClassifier and print a test metric

`DummyClassifier` is a baseline: it does not learn from the features. **A high score is not the goal.** We only need proof that sklearn can fit and score on your split.


```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

# Ready to run.
clf = DummyClassifier(strategy="most_frequent")
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

# A high score is not the goal — this is a baseline, not a model you are trying to beat.
test_acc = accuracy_score(y_test, y_pred)
print("test accuracy (DummyClassifier, most_frequent):", test_acc)
```
